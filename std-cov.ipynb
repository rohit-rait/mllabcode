```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
X = [2, 4, 6, 8, 10]
Y = [1, 3, 7, 11, 14]
```

```python
X_np = np.array(X)
Y_np = np.array(Y)
```

```python
mean_X = np.mean(X_np)
mean_Y = np.mean(Y_np)
```

```python
std_X = np.std(X_np, ddof=1)
std_Y = np.std(Y_np, ddof=1)
```

```python
larger_spread = 'X' if std_X > std_Y else 'Y'
```

```python
covariance = np.cov(X_np, Y_np)[0, 1]
```

```python
print(f"Mean of X: {mean_X}")
print(f"Mean of Y: {mean_Y}")
print(f"Standard Deviation of X: {std_X}")
print(f"Standard Deviation of Y: {std_Y}")
print(f"Larger spread is in dataset: {larger_spread}")
print(f"Covariance of X: {math.sqrt(std_X)}");
print(f"Covariance of Y: {math.sqrt(std_Y)}");
print(f"Covariance between X and Y: {covariance}")
```

```python
plt.figure(figsize=(12, 5))
```

```python
plt.subplot(1, 2, 1)
plt.plot(X_np, label='X', marker='o')
plt.plot(Y_np, label='Y', marker='s')
plt.title('Line Plot of X and Y')
plt.xlabel('Index')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
```

```python
plt.subplot(1, 2, 2)
plt.scatter(X_np, Y_np, c='green', marker='x')
plt.title('Scatter Plot (X vs Y)')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
```

```python
plt.tight_layout()
plt.show()
```
