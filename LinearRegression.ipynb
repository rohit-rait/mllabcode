```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Sample data: Age of car (in years) and corresponding speed (in mph)
age_of_car = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])  # x
speed_of_car = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])  # y
```

```python
# Step 1: Scatter plot of the data
plt.figure(figsize=(10, 6))  # Set the figure size
plt.scatter(age_of_car, speed_of_car, color='blue', label='Data points')
plt.title('Scatter Plot of Car Age vs Speed')
plt.xlabel('Age of Car (years)')
plt.ylabel('Speed of Car (mph)')
plt.legend()
plt.grid()
plt.show()  # Show the scatter plot before regression calculations
```

```python
# Reshape the data for sklearn
X = age_of_car.reshape(-1, 1)  # Reshape for sklearn
y = speed_of_car
```

```python
# Step 2: Linear regression: Speed as a function of Age
model_xy = LinearRegression()
model_xy.fit(X, y)
```

```python
predicted_speed = model_xy.predict(X)
```

```python
age_to_predict = float(input("Enter the age of the car (in years) to predict its speed: "))
predicted_speed_value = model_xy.predict(np.array([[age_to_predict]]))
print(f"Predicted speed of a {age_to_predict}-year-old car: {predicted_speed_value[0]:.2f} mph")
```

```python
model_yx = LinearRegression()
model_yx.fit(y.reshape(-1, 1), X)
```

```python
predicted_age = model_yx.predict(y.reshape(-1, 1))
```

```python
plt.figure(figsize=(10, 6))  # Set the figure size for the new plot
plt.scatter(age_of_car, speed_of_car, color='blue', label='Data points')
plt.plot(age_of_car, predicted_speed, color='red', label='Regression Line (Speed on Age)')
plt.plot(predicted_age, y, color='green', label='Regression Line (Age on Speed)')
plt.title('Scatter Plot with Regression Lines')
plt.xlabel('Age of Car (years)')
plt.ylabel('Speed of Car (mph)')
plt.legend()
plt.grid()
```

```python
slope_xy = model_xy.coef_[0]
intercept_xy = model_xy.intercept_
print(f"Equation of Speed as a function of Age: Speed = {slope_xy:.2f} * Age + {intercept_xy:.2f}")
```

```python
slope_yx = model_yx.coef_[0][0]
intercept_yx = model_yx.intercept_[0]
print(f"Equation of Age as a function of Speed: Age = {slope_yx:.2f} * Speed + {intercept_yx:.2f}")
```
