```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
```

```python
print("Target Name:", data.target_names)
print("\nDataset Features (First 5 rows):\n", pd.DataFrame(data.data, columns=data.feature_names).head())
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data.data)
```

```python
pca = PCA(n_components=3)
pca_components = pca.fit_transform(scaled_data)
```

```python
print("\nExplained Variance Ratio:", pca.explained_variance_ratio_)
print("\nPCA Components (Principal Axes):\n", pca.components_)
```

```python
plt.figure(figsize=(8, 6))
scatter = plt.scatter(pca_components[:, 0], pca_components[:, 1], c=data.target, cmap='viridis')
plt.title('PCA1 vs PCA2')
plt.xlabel('PCA1')
plt.ylabel('PCA2')
plt.colorbar(scatter, label='Target')
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D
```

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(pca_components[:, 0], pca_components[:, 1], pca_components[:, 2], c=data.target, cmap='viridis')
ax.set_xlabel('PCA1')
ax.set_ylabel('PCA2')
ax.set_zlabel('PCA3')
ax.set_title('3D PCA: PCA1, PCA2, and PCA3')

# Add color bar
fig.colorbar(scatter, ax=ax, label='Target')

plt.show()
```
